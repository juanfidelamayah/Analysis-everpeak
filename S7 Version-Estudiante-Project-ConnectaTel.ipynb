## 🧩Paso 3: Limpieza básica de datos

### 3.1 Corregir sentinels y fechas imposibles
**🎯 Objetivo:**  
Aplicar reglas de limpieza para reemplazar valores sentinels y corregir fechas imposibles.

**Instrucciones:**  
- En `age`, reemplaza el sentinel **-999** con la mediana.
- En `city`, reemplaza el sentinel `"?"` por valores nulos (`pd.NA`).  
- Marca como nulas (`pd.NA`) las fechas fuera de rango.

```python
# Convertir age a tipo numérico (por si acaso hay texto o valores raros)
users['age'] = pd.to_numeric(users['age'], errors='coerce')

# Calcular la mediana únicamente para valores de edad válidos (mayores a 0)
age_mediana = users[users['age'] > 0]['age'].median()

# Reemplazar los valores -999 (o nulos) por la mediana
users['age'] = users['age'].replace(-999, age_mediana).fillna(age_mediana)

# Verificar los cambios
users['age'].describe()
```

```python
# Reemplazar ? por NA en city
users['city'] = users['city'].replace('?', pd.NA)

# Verificar cambios
users['city'].value_counts(dropna=False)
```

```python
# Marcar fechas futuras como NA para reg_date
users.loc[users['reg_date'].dt.year > 2024, 'reg_date'] = pd.NaT

# Verificar cambios
users['reg_date'].dt.year.value_counts(dropna=False)
```

## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico. 

**Instrucciones:**: 
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python
# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int) #conocer el total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int) #conocer el total de llamadas


# Agrupar información por usuario
usage_agg = usage.groupby("user_id").agg(
    cant_mensajes=("is_text", "sum"),
    cant_llamadas=("is_call", "sum"),
    cant_minutos_llamada=("duration", "sum")
).reset_index()

# observar resultado
usage_agg.head(3)
```

```python
# Renombrar columnas
usage_agg = usage_agg.rename(columns={
    'cant_mensajes': 'total_messages',
    'cant_llamadas': 'total_calls',
    'cant_minutos_llamada': 'total_call_minutes'
})

# observar resultado
usage_agg.head(3)
```

```python
# Combinar la tabla agregada con el dataset de usuarios
user_profile = users.merge(usage_agg, on='user_id', how='left')

# Rellenar con 0 a los usuarios sin registros de uso
user_profile[['total_messages', 'total_calls', 'total_call_minutes']] = user_profile[['total_messages', 'total_calls', 'total_call_minutes']].fillna(0)

user_profile.head(5)
```

**4. Boxplot: total_call_minutes (Total Minutos de Llamada)Presencia de outliers:** Presenta outliers.Análisis: Se observan múltiples puntos atípicos por encima de la cola superior, especialmente notables en el plan Premium. La mediana de minutos hablados es mayor en el plan Premium, y sus outliers alcanzan las mayores acumulaciones de minutos consumidos en el período analizado.

```python
# Calcular límites con el método IQR
columnas_limites = ['age', 'total_messages', 'total_calls', 'total_call_minutes']

for col in columnas_limites:
    Q1 = user_profile[col].quantile(0.25)
    Q3 = user_profile[col].quantile(0.75)
    IQR = Q3 - Q1
    
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    
    print(f"--- Límites IQR para '{col}' ---")
    print(f"Q1: {Q1:.2f} | Q3: {Q3:.2f} | IQR: {IQR:.2f}")
    print(f"Límite Inferior: {limite_inferior:.2f}")
    print(f"Límite Superior: {limite_superior:.2f}\n")
```

```python
# Revisa los limites superiores y el max, para tomar la decisión de mantener los outliers o no
user_profile[columnas_limites].describe()
```

💡Insights: 
**- cant_mensajes (total_messages):** Se deben mantener. Los valores atípicos corresponden a usuarios de alto tráfico (heavy users) cuyo comportamiento de envío masivo de mensajes es real e inherente al servicio. Eliminarlos sesgaría la comprensión del uso del plan y el cálculo de ingresos por consumos adicionales.

**- cant_llamadas (total_calls):** Se recomienda mantener. Reflejan la variabilidad natural del comportamiento del cliente. Una mayor frecuencia de llamadas es esperada y válida dentro del perfil de ciertos usuarios, especialmente en el plan Premium.

**cant_minutos_llamada (total_call_minutes):** Se deben Mantener. Los minutos acumulados representan consumo efectivo de la red. Retener estos datos es fundamental para dimensionar correctamente la demanda operativa de infraestructura y evaluar la rentabilidad real de los planes tarifarios.

## 🧩Paso 6: Segmentación de Clientes

### 6.1 Segmentación de Clientes Por Uso

🎯 **Objetivo:** Clasificar a cada usuario en un grupo de uso (Bajo uso, Uso medio, Alto uso) basándose en la cantidad de llamadas y mensajes registrados.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_uso` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python
# Crear columna grupo_uso
import numpy as np

# Definir las condiciones evaluando llamadas y mensajes
condiciones = [
    (user_profile['total_calls'] < 5) & (user_profile['total_messages'] < 5),
    (user_profile['total_calls'] < 10) & (user_profile['total_messages'] < 10)
]

# Definir las etiquetas de los dos primeros grupos
etiquetas = ['Bajo uso', 'Uso medio']

# Asignar etiquetas y 'Alto uso' para todos los casos restantes
user_profile['grupo_uso'] = np.select(condiciones, etiquetas, default='Alto uso')

# Verificar el resultado con un conteo
user_profile['grupo_uso'].value_counts()
```

### 6.2 Segmentación de Clientes Por Edad

🎯 **Objetivo:**: Clasificar a cada usuario en un grupo por **edad**.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_edad` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones y asigna:
  - `'Joven'` cuando age < 30
  - `'Adulto'` cuando age < 60
  - `'Adulto Mayor'` para el resto de casos

```python
# Crear columna grupo_edad
import numpy as np

# Definir las condiciones evaluando la edad
condiciones_edad = [
    user_profile['age'] < 30,
    user_profile['age'] < 60
]

# Definir las etiquetas de los dos primeros grupos
etiquetas_edad = ['Joven', 'Adulto']

# Asignar etiquetas y 'Adulto Mayor' para los casos restantes
user_profile['grupo_edad'] = np.select(condiciones_edad, etiquetas_edad, default='Adulto Mayor')

# Verificar el resultado con la distribución de usuarios por grupo
user_profile['grupo_edad'].value_counts()
```
